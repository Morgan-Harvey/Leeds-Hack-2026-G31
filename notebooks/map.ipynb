```python
import pandas as pd
import folium

# Load your CSV
df = pd.read_csv("../datasets/raw/London stations.csv")

# Create a base map centred roughly on London
m = folium.Map(location=[51.5074, -0.1278], zoom_start=11)

# Loop through each station and add a dot
for _, row in df.iterrows():
    folium.CircleMarker(
        location=[row["Latitude"], row["Longitude"]],
        radius=4,
        color="blue",
        fill=True,
        fill_opacity=0.7,
        popup=row["Station"]
    ).add_to(m)

m
```

```python
import pandas as pd
import folium

df = pd.read_csv("../datasets/raw/London stations.csv")

# Drop rows where Zone is missing
df = df.dropna(subset=["Zone"])

# Extract the first zone (handles "3" and "3,4,5,6")
df["Zone"] = df["Zone"].astype(str).str.split(",").str[0].astype(int)

zone_colors = {
    1: "red",
    2: "orange",
    3: "green",
    4: "blue",
    5: "purple",
    6: "pink",
    7: "brown"
}

m = folium.Map(location=[51.5074, -0.1278], zoom_start=11)

for _, row in df.iterrows():
    zone = row["Zone"]
    color = zone_colors.get(zone, "black")

    folium.CircleMarker(
        location=[row["Latitude"], row["Longitude"]],
        radius=5,
        color=color,
        fill=True,
        fill_color=color,
        fill_opacity=0.8,
        popup=f"{row['Station']} (Zone {zone})"
    ).add_to(m)

m
```

```python
import matplotlib.cm as cm
import matplotlib.colors as colors

def scalar_to_hex(x):
    cmap = cm.get_cmap("viridis")   # or plasma, inferno, magma, etc.
    rgba = cmap(x)                  # x must be between 0 and 1
    return colors.to_hex(rgba)
```
