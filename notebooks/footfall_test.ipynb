```python
import pandas as pd


df = pd.read_csv('../datasets/raw/StationFootfall_2024_2025.csv')

df.rename(columns={'TravelDate' : 'date', 'DayOfWeek' : 'day_of_week', 'Station' : 'station','EntryTapCount' : 'entry_tap_count', 'ExitTapCount' : 'exit_tap_count'}, inplace=True)

df["date"] = pd.to_datetime(df["date"].astype(str), format="%Y%m%d")

df["baseline_entry"] = (
    df.groupby(["station", "day_of_week"])["entry_tap_count"]
      .transform("mean")
)

df["baseline_exit"] = (
    df.groupby(["station", "day_of_week"])["exit_tap_count"]
      .transform("mean")
)

df["overcrowding_index"] = (
    df["entry_tap_count"] / df["baseline_entry"]
) * 100

print(df.head())
```
